```python
import pandas as pd
import numpy as np
import lightgbm as lgb
from sklearn.metrics import roc_auc_score, accuracy_score

# =========================
# 1) datetime + 기본 전처리
# =========================
df = df.copy()

df["review_dt"] = pd.to_datetime(df["timestamp_created"], unit="s", errors="coerce")
df = df.dropna(subset=["review_dt"]).copy()

# review_length
df["review_length"] = df["review"].fillna("").astype(str).str.len()

# deck_playtime_at_review 결측 처리 (컬럼 있으면)
if "deck_playtime_at_review" in df.columns:
    df["deck_playtime_at_review"] = df["deck_playtime_at_review"].fillna(0)

# True/False -> 0/1 정리 (LightGBM 편하게)
bool_cols = [
    "primarily_steam_deck",
    "voted_up",
    "steam_purchase",
    "received_for_free",
    "written_during_early_access",
]
for c in bool_cols:
    if c in df.columns:
        df[c] = df[c].astype(int)

# =========================
# 2) 180일 중 마지막 90일 제외 (라벨 생성용 구간만)
# =========================
END_DATE = df["review_dt"].max()
START_DATE = END_DATE - pd.Timedelta(days=180)
LABEL_CUTOFF = END_DATE - pd.Timedelta(days=90)

df_180 = df[df["review_dt"] >= START_DATE].copy()
df_label = df_180[df_180["review_dt"] <= LABEL_CUTOFF].copy()

# =========================
# 3) 15/30/60 Stage 라벨 생성
# - last_played는 피처로 쓰지 않지만, 라벨 생성엔 사용
# - stage 정의:
#   0: 0–14일
#   1: 15–29일
#   2: 30–59일
#   3: 60일 이상 (또는 관측 없음/이상치 포함)
# =========================
df_label["last_played_dt"] = pd.to_datetime(df_label["last_played"], unit="s", errors="coerce")

# days_after_review 계산
df_label["days_after_review"] = (df_label["last_played_dt"] - df_label["review_dt"]).dt.days

# 기본 stage=3 (장기/미관측)
df_label["stage"] = 3

# last_played가 0이거나 NaT면 "리뷰 이후 플레이 없음" -> stage 0으로 둘지 3으로 둘지 정책 필요
# 너가 기존에 churn=1로 잡던 방향(=이탈)과 비슷하게 하려면 "가장 빠른 이탈"로 두는게 보통 더 일관적이야.
no_play_mask = (df_label["last_played"] == 0) | (df_label["last_played_dt"].isna())
df_label.loc[no_play_mask, "stage"] = 0

# 날짜 이상치(리뷰보다 과거에 last_played가 찍힘)도 빠른 이탈/데이터 오류로 보고 0 처리
df_label.loc[df_label["days_after_review"] < 0, "stage"] = 0

# 정상 days_after_review에 대해 stage 부여
m = (~no_play_mask) & (df_label["days_after_review"] >= 0)
df_label.loc[m & (df_label["days_after_review"] < 15), "stage"] = 0
df_label.loc[m & (df_label["days_after_review"] >= 15) & (df_label["days_after_review"] < 30), "stage"] = 1
df_label.loc[m & (df_label["days_after_review"] >= 30) & (df_label["days_after_review"] < 60), "stage"] = 2
df_label.loc[m & (df_label["days_after_review"] >= 60), "stage"] = 3

# =========================
# 4) 피처 선택 (사용자가 준 리스트 그대로)
# =========================
features = [
    "num_games_owned",
    "num_reviews_author",
    "deck_playtime_at_review",
    "voted_up",
    "votes_up",
    "votes_funny",
    "weighted_vote_score",
    "comment_count",
    "steam_purchase",
    "received_for_free",
    "written_during_early_access",
    "review_length",
]

features = [c for c in features if c in df_label.columns]

for c in features:
    df_label[c] = pd.to_numeric(df_label[c], errors="coerce")

X = df_label[features].fillna(0)
y = df_label["stage"].astype(int)

# =========================
# 5) 시간 기준 Train/Valid Split (라벨 구간의 마지막 30일을 valid)
# =========================
split_date = LABEL_CUTOFF - pd.Timedelta(days=30)

train_mask = df_label["review_dt"] <= split_date
valid_mask = df_label["review_dt"] > split_date

X_train, y_train = X[train_mask], y[train_mask]
X_valid, y_valid = X[valid_mask], y[valid_mask]

print("Rows:", len(df_label), "| Train:", len(X_train), "| Valid:", len(X_valid))
print("Stage rate train:", (y_train.value_counts(normalize=True).sort_index().round(4).to_dict()))
print("Stage rate valid:", (y_valid.value_counts(normalize=True).sort_index().round(4).to_dict()))
print("Features used:", features)

# valid에 클래스가 너무 부족하면(특정 stage가 0개) AUC macro 계산이 불안정할 수 있음
if y_valid.nunique() < 2:
    raise ValueError(f"Valid set에 클래스가 1개뿐입니다. (unique={sorted(y_valid.unique())}) split_date를 조정하거나 기간을 늘려야 합니다.")

# =========================
# 6) LightGBM 멀티클래스 학습 + 평가
# =========================
NUM_CLASS = 4

model = lgb.LGBMClassifier(
    objective="multiclass",
    num_class=NUM_CLASS,
    n_estimators=1200,
    learning_rate=0.05,
    num_leaves=63,
    subsample=0.8,
    colsample_bytree=0.8,
    random_state=42
)

model.fit(
    X_train, y_train,
    eval_set=[(X_valid, y_valid)],
    eval_metric="multi_logloss"
)

proba = model.predict_proba(X_valid)  # (N, 4)
pred_class = np.argmax(proba, axis=1)

acc = accuracy_score(y_valid, pred_class)

# 멀티클래스 AUC (OVR, macro)
# - y_true는 (N,) 레이블, y_score는 (N, C) 확률
# - valid에 없는 클래스가 있으면 에러/불안정할 수 있어 위에서 비율 출력 꼭 확인
auc_macro_ovr = roc_auc_score(y_valid, proba, multi_class="ovr", average="macro")

print(f"Validation ACC: {acc:.4f}")
print(f"Validation ROC-AUC (macro OVR): {auc_macro_ovr:.4f}")

# =========================
# 7) (선택) 최근 90일(inference 구간) stage 확률 예측
# =========================
df_recent = df_180[df_180["review_dt"] > LABEL_CUTOFF].copy()

df_recent["review_length"] = df_recent["review"].fillna("").astype(str).str.len()
if "deck_playtime_at_review" in df_recent.columns:
    df_recent["deck_playtime_at_review"] = df_recent["deck_playtime_at_review"].fillna(0)
for c in bool_cols:
    if c in df_recent.columns:
        df_recent[c] = df_recent[c].astype(int)
for c in features:
    df_recent[c] = pd.to_numeric(df_recent[c], errors="coerce")

X_recent = df_recent[features].fillna(0)
proba_recent = model.predict_proba(X_recent)

# stage별 확률 컬럼 추가
for k in range(NUM_CLASS):
    df_recent[f"stage{k}_prob"] = proba_recent[:, k]

# 예: "빠른 이탈(stage0)" 확률을 위험도로 쓰고 싶으면 이걸 쓰면 됨
df_recent["risk_prob"] = df_recent["stage0_prob"]

df_recent[["steamid", "appid", "review_dt", "risk_prob", "stage0_prob", "stage1_prob", "stage2_prob", "stage3_prob"]].head()
```

```python
import numpy as np

from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.svm import LinearSVC
from sklearn.calibration import CalibratedClassifierCV

from sklearn.metrics import (
    roc_auc_score, average_precision_score,
    classification_report, confusion_matrix
)

def train_eval_linear_svm(
    X_train, y_train,
    X_valid, y_valid,
    C=1.0,
    calibrate=True,
    calib_method="sigmoid",   # "sigmoid" or "isotonic"
    calib_cv=3,
    threshold=0.5
):
    """
    Linear SVM 베이스라인:
    - calibrate=True면 확률 추정(Platt scaling)으로 AUC/PR-AUC 계산이 안정적
    - threshold로 최종 라벨 예측(기본 0.5)
    """

    base = Pipeline([
        # 희소행렬(TF-IDF 등)도 지원하려고 with_mean=False
        ("scaler", StandardScaler(with_mean=False)),
        ("svm", LinearSVC(C=C, class_weight="balanced", random_state=42))
    ])

    if calibrate:
        # LinearSVC는 predict_proba가 없어서 CalibratedClassifierCV로 확률 만들기
        clf = CalibratedClassifierCV(base, method=calib_method, cv=calib_cv)
    else:
        clf = base

    clf.fit(X_train, y_train)

    # 점수/확률 얻기
    if calibrate:
        y_score = clf.predict_proba(X_valid)[:, 1]   # 확률
    else:
        # calibration 안 하면 decision_function을 점수로 사용 (AUC 계산 가능)
        y_score = clf.decision_function(X_valid)

    # AUC 계열
    auc = roc_auc_score(y_valid, y_score)
    pr_auc = average_precision_score(y_valid, y_score)

    # threshold 기반 예측
    if calibrate:
        y_pred = (y_score >= threshold).astype(int)
    else:
        # calibration 없을 땐 decision score를 0 기준으로 라벨링하는 게 일반적
        y_pred = (y_score >= 0).astype(int)

    print(f"[Linear SVM] C={C} | calibrate={calibrate} ({calib_method})")
    print(f"ROC-AUC: {auc:.4f}")
    print(f"PR-AUC : {pr_auc:.4f}\n")

    print("Confusion matrix:")
    print(confusion_matrix(y_valid, y_pred), "\n")

    print("Classification report:")
    print(classification_report(y_valid, y_pred, digits=4))

    return clf, {"roc_auc": auc, "pr_auc": pr_auc}


#===== 사용 예시 =====
#X_train, y_train, X_valid, y_valid 가 이미 준비돼 있다고 가정
#(너는 LightGBM 돌렸던 그대로 같은 split을 쓰는 게 핵심)

#1) 기본 1회 실행
clf, metrics = train_eval_linear_svm(X_train, y_train, X_valid, y_valid, C=1.0, calibrate=True)

#2) C 몇 개만 간단 스윕
for C in [0.01, 0.1, 1.0, 5.0, 10.0]:
    _clf, _m = train_eval_linear_svm(X_train, y_train, X_valid, y_valid, C=C, calibrate=True)
```
