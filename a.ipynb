# 결측치 처리

```python
df.isnull().sum()
```

```python
# 이유: 이벤트 타입이 없으면 그 row는 해석 자체가 불가능(데이터 오류에 가까움)
df = df.dropna(subset=["behav_comptype_h"])
```

```python
# 이유: Production Director처럼 상사가 없는 경우를 범주로 명시하는 게 모델에 유리함.
sup_cat = ["sup_role", "sup_sex", "sup_fname", "sup_lname"]
df[sup_cat] = df[sup_cat].fillna("None")
```

```python
# 이유: “상사 없음”을 **명시(플래그)**하면서 수치형 결측으로 인한 모델 오류를 막음.
df["has_supervisor"] = df["sup_ID"].notna().astype(int)
sup_num = ["sup_ID", "sup_age", "sup_sub_age_diff",
           "sup_commitment_h", "sup_perceptiveness_h", "sup_goodness_h"]
df[sup_num] = df[sup_num].fillna(-1)
```

```python
# 이유: 동료가 없는 직책/상황을 구분하는 신호가 됨.
df["has_collaborators"] = df["sub_coll_IDs"].notna().astype(int)
df["sub_coll_IDs"] = df["sub_coll_IDs"].fillna("None")
```

```python
# 이유: 동료가 없으면 비율 자체가 정의 불가이므로 “없음(-1)”로 분리.
df["has_collaborators_prtn"] = df["sub_colls_same_sex_prtn"].notna().astype(int)
df["sub_colls_same_sex_prtn"] = df["sub_colls_same_sex_prtn"].fillna(-1)
```

behav_cause_h (거의 전부 결측)
이유: Resignation일 때만 값이 있고, 나머지는 원래 None이 정상.

actual_efficacy_h (결측 많음)
이유: Efficacy 이벤트에서만 의미가 있는 값이라 다른 이벤트에서는 비어야 정상.

record_comptype (결측 3,938)
이유: 상사가 기록을 안 남긴 이벤트는 None이 정상(“기록 없음” 자체가 의미).

record_cause (거의 전부 결측)
이유: Termination일 때만 존재하는 사유라 나머지 결측은 정상.

recorded_efficacy (결측 많음)
이유: Efficacy 기록이 있는 이벤트에서만 존재(다른 이벤트에서 채우면 왜곡).

recorded_note_from_sup (결측 매우 많음)
이유: 일부 정성 기록(Teamwork/Feat 등)에서만 메모가 달리며 대부분 None이 정상.

record_conf_matrix_h (결측 매우 많음)
이유: 정성 이벤트에서 “기록이 정확했는지”를 나타내는 값이라 해당 이벤트가 아니면 None이 정상.
