```python
import torch
```

```python
# 역전파를 위해서 x에 대한 gradient 계산 활성
x = torch.FloatTensor([[1,2],
                       [3,4]]).requires_grad_(True)
```

```python
x
```

```python
# 연산트리 자동 생성
x1 = x + 2              # 모든 원소에 + 2
x2 = x - 2              # 모든 원소에 - 2
x3 = x1 * x2            # (x+2)와 (x-2) 텐서를 원소별로 곱함
y = x3.sum()            # 모든 원소를 더해 스칼라값으로 변환

print(x1)
print(x2)
print(x3)
print(y)
```

```python
# 자동 미분을 통한 역전파
y.backward() 
print(x.grad)
```

```python
x3.detach_().numpy()    # inplace 계산그래프 분리
```
