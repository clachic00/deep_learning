```python
import pandas as pd

def make_daily_panel(df: pd.DataFrame) -> pd.DataFrame:
    df = df.copy()

    key = ["sub_ID", "event_day_in_series"]

    # (A) 개인/상사/팀/시간: "그날의 속성" (대부분 고정/동일) → first
    base_cols = [
        "sub_age","sub_sex","sub_role","sub_shift","sub_team",
        # "sub_health_h","sub_commitment_h","sub_perceptiveness_h","sub_dexterity_h",
        # "sub_sociality_h","sub_goodness_h","sub_strength_h","sub_openmindedness_h",
        "has_supervisor","sup_age","sup_sub_age_diff","sup_sex","sup_role",
        # "sup_commitment_h","sup_perceptiveness_h","sup_goodness_h",
        "has_collaborators","sub_colls_same_sex_prtn","has_collaborators_prtn",
        "event_week_in_series","event_weekday_num",
    ]
    base = (df[key + base_cols]
            .sort_values(key)
            .groupby(key, as_index=False)
            .first())

    # (B) 이벤트 타입별 카운트 → pivot
    # 예: cnt_Presence, cnt_Absence, cnt_Disruption ...
    evt = df[key + ["behav_comptype_h"]].dropna()
    evt_cnt = (evt
               .assign(cnt=1)
               .pivot_table(index=key, columns="behav_comptype_h", values="cnt",
                            aggfunc="sum", fill_value=0)
               .reset_index())
    evt_cnt.columns = [c if c in key else f"cnt_{c}" for c in evt_cnt.columns]

    #  efficacy 요약 (있는 날만 평균) → 없으면 NaN
    eff = (df[key + ["actual_efficacy_h"]]
           .groupby(key, as_index=False)["actual_efficacy_h"]
           .mean()
           .rename(columns={"actual_efficacy_h": "eff_mean_today"}))

    # (D) 합치기
    panel = base.merge(evt_cnt, on=key, how="left").merge(eff, on=key, how="left")

    # 이벤트 카운트가 없는 날은 0으로
    cnt_cols = [c for c in panel.columns if c.startswith("cnt_")]
    panel[cnt_cols] = panel[cnt_cols].fillna(0).astype(int)

    return panel


# 사용
panel = make_daily_panel(df)
panel.head()
```

```python
import pandas as pd
import numpy as np

def build_panel_and_rolling_features(df: pd.DataFrame, windows=(7, 14, 30)) -> pd.DataFrame:
    df = df.copy()
    key = ["sub_ID", "event_day_in_series"]

    # -------------------------
    # 1) "사람×날짜" 패널 만들기 (1일 1행)
    # -------------------------
    base_cols = [
        "sub_age","sub_sex","sub_role","sub_shift","sub_team",
        # "sub_health_h","sub_commitment_h","sub_perceptiveness_h","sub_dexterity_h",
        # "sub_sociality_h","sub_goodness_h","sub_strength_h","sub_openmindedness_h",
        "has_supervisor","sup_age","sup_sub_age_diff","sup_sex","sup_role",
        # "sup_commitment_h","sup_perceptiveness_h","sup_goodness_h",
        "has_collaborators","sub_colls_same_sex_prtn","has_collaborators_prtn",
        "event_week_in_series","event_weekday_num"
    ]

    base = (df.sort_values(key)
              .groupby(key, as_index=False)[base_cols]
              .first())

    # 이벤트 타입별 하루 카운트 (cnt_Presence, cnt_Absence, ...)
    evt_cnt = (df[key + ["behav_comptype_h"]]
                .dropna()
                .assign(cnt=1)
                .pivot_table(index=key, columns="behav_comptype_h",
                             values="cnt", aggfunc="sum", fill_value=0)
                .reset_index())
    evt_cnt.columns = [c if c in key else f"cnt_{c}" for c in evt_cnt.columns]

    # 하루 actual/recorded efficacy 평균
    eff = (df[key + ["actual_efficacy_h", "recorded_efficacy"]]
            .groupby(key, as_index=False)
            .agg(
                eff_actual_today=("actual_efficacy_h", "mean"),
                eff_recorded_today=("recorded_efficacy", "mean")
            ))

    # 격차(절대값) - 둘 다 있는 날만 계산되도록
    eff["eff_gap_today"] = (eff["eff_actual_today"] - eff["eff_recorded_today"]).abs()

    panel = (base.merge(evt_cnt, on=key, how="left")
                 .merge(eff, on=key, how="left"))

    # 이벤트 카운트 없는 날은 0
    cnt_cols = [c for c in panel.columns if c.startswith("cnt_")]
    panel[cnt_cols] = panel[cnt_cols].fillna(0).astype(int)

    # -------------------------
    # 2) Rolling 집계 피처 만들기 (누출 방지: shift(1))
    # -------------------------
    panel = panel.sort_values(["sub_ID", "event_day_in_series"]).reset_index(drop=True)

    # 최소 추천 이벤트들만 고정해서 뽑기 (없으면 0컬럼 생성)
    must_events = ["Presence", "Absence", "Disruption", "Teamwork"]
    for e in must_events:
        col = f"cnt_{e}"
        if col not in panel.columns:
            panel[col] = 0

    # 효율 값이 너무 큰 이상치가 있으면(너가 본 max 3~4대) 캡 걸어도 됨 (선택)
    # panel["eff_actual_today"]  = panel["eff_actual_today"].clip(upper=2.0)
    # panel["eff_recorded_today"] = panel["eff_recorded_today"].clip(upper=2.0)
    # panel["eff_gap_today"]     = panel["eff_gap_today"].clip(upper=2.0)

    # rolling 계산용 함수: 과거만 보도록 shift(1) 후 rolling
    def roll_sum(s, w):  return s.shift(1).rolling(w, min_periods=1).sum()
    def roll_mean(s, w): return s.shift(1).rolling(w, min_periods=1).mean()
    def roll_std(s, w):  return s.shift(1).rolling(w, min_periods=2).std()

    # (A) 이벤트 카운트: 7/14/30일 합
    for w in windows:
        for e in must_events:
            src = f"cnt_{e}"
            panel[f"{src}_sum_{w}d"] = panel.groupby("sub_ID")[src].transform(lambda s: roll_sum(s, w)).astype(float)

    # (B) efficacy: 평균/표준편차
    for w in windows:
        panel[f"eff_actual_mean_{w}d"] = panel.groupby("sub_ID")["eff_actual_today"].transform(lambda s: roll_mean(s, w))
        panel[f"eff_actual_std_{w}d"]  = panel.groupby("sub_ID")["eff_actual_today"].transform(lambda s: roll_std(s, w))

        panel[f"eff_gap_mean_{w}d"]    = panel.groupby("sub_ID")["eff_gap_today"].transform(lambda s: roll_mean(s, w))
        panel[f"eff_gap_std_{w}d"]     = panel.groupby("sub_ID")["eff_gap_today"].transform(lambda s: roll_std(s, w))

    # 표준편차 NaN(과거 1개뿐인 구간) → 0으로(안정성 피처로 해석)
    std_cols = [c for c in panel.columns if c.endswith("_std_7d") or c.endswith("_std_14d") or c.endswith("_std_30d")]
    panel[std_cols] = panel[std_cols].fillna(0)

    # (선택) 변화량/추세: 단기-장기 차이 (점수 도움될 때 많음)
    if 7 in windows and 30 in windows:
        panel["eff_actual_trend_7v30"] = panel["eff_actual_mean_7d"] - panel["eff_actual_mean_30d"]
        panel["abs_trend_7v30"]        = panel["cnt_Absence_sum_7d"] - panel["cnt_Absence_sum_30d"]

    return panel



panel = build_panel_and_rolling_features(df, windows=(7,14,30))
panel.head()
```

```python
import pandas as pd
import numpy as np

def add_resignation_label(panel: pd.DataFrame, horizon: int = 60) -> pd.DataFrame:
    """
    panel: 1일 1행(= sub_ID, event_day_in_series 유니크) 패널 데이터
    horizon: t 이후 horizon일 내 Resignation이면 y=1
    
    생성:
      - resignation_day: 개인별 첫 Resignation 발생 day
      - y: (t < resignation_day <= t+horizon)
    제거:
      - t >= resignation_day 인 행(이탈 이후/당일) 제거 (누출 방지)
    """
    panel = panel.copy()

    # 1) 개인별 resignation_day 계산 (원본 패널에 resignation_day가 없다고 가정)
    #    -> panel에 "resignation_day"가 이미 있으면 그대로 사용 가능
    if "resignation_day" not in panel.columns:
        # panel에 resignation 이벤트 표시 컬럼이 있는 경우를 우선 지원
        # (앞에서 이벤트 카운트 만들었으면 cnt_Resignation이 있을 수 있음)
        if "cnt_Resignation" in panel.columns:
            res_day = (panel.loc[panel["cnt_Resignation"] > 0, ["sub_ID", "event_day_in_series"]]
                            .groupby("sub_ID")["event_day_in_series"]
                            .min())
        else:
            raise ValueError("panel에 'resignation_day'도 없고, 'cnt_Resignation'도 없습니다. "
                             "원본 로그(df)에서 resignation_day를 먼저 구해 merge하거나, "
                             "패널 생성 시 cnt_Resignation을 포함하세요.")

        panel = panel.merge(res_day.rename("resignation_day"), on="sub_ID", how="left")

    # 2) 라벨 생성: t < resignation_day <= t+horizon
    t = panel["event_day_in_series"]
    rd = panel["resignation_day"]

    panel["y"] = ((rd.notna()) & (t < rd) & (rd <= t + horizon)).astype(int)

    # 3) 누출 방지: 이탈 당일/이후 제거 (t >= resignation_day)
    panel = panel[(rd.isna()) | (t < rd)].copy()

    return panel


# 사용
panel = build_panel_and_rolling_features(df)  # 이전에 만든 패널+파생피처
panel = add_resignation_label(panel, horizon=60)
panel["y"].value_counts()
```

```python
import numpy as np
import pandas as pd
from sklearn.metrics import roc_auc_score, precision_score, recall_score, f1_score
from lightgbm import LGBMClassifier


def train_eval_auc(
    panel_df,
    features,
    label_col="y",
    time_col="event_day_in_series",
    split_day=None,
    valid_days=60,
    random_state=7777,
    threshold=0.5,
    topk_ratio=0.01,
    # ===== 추가: threshold sweep =====
    sweep_thresholds=True,
    t_min=0.01,
    t_max=0.30,
    t_steps=30,
):
    df = panel_df.copy()

    # 0) feature 존재 확인
    missing = [c for c in features + [label_col, time_col] if c not in df.columns]
    if missing:
        raise ValueError(f"Missing columns: {missing}")

    # 1) 시간 split
    if split_day is None:
        max_day = int(df[time_col].max())
        split_day = max_day - valid_days

    train = df.loc[df[time_col] <= split_day].copy()
    valid = df.loc[df[time_col] >  split_day].copy()

    # 2) X / y
    X_train = train.loc[:, features].copy()
    y_train = train.loc[:, label_col].astype(int).to_numpy()

    X_valid = valid.loc[:, features].copy()
    y_valid = valid.loc[:, label_col].astype(int).to_numpy()

    # 3) categorical -> one-hot
    cat_cols = [c for c in X_train.columns if X_train[c].dtype == "object"]
    X_train = pd.get_dummies(X_train, columns=cat_cols, dummy_na=True)
    X_valid = pd.get_dummies(X_valid, columns=cat_cols, dummy_na=True)

    # train/valid 컬럼 정렬 맞추기
    X_valid = X_valid.reindex(columns=X_train.columns, fill_value=0)

    # 4) imbalance
    y0 = int((y_train == 0).sum())
    y1 = int((y_train == 1).sum())
    scale_pos_weight = y0 / max(y1, 1)

    # 5) model
    model = LGBMClassifier(
        objective="binary",
        n_estimators=1500,
        learning_rate=0.03,
        num_leaves=63,
        subsample=0.8,
        colsample_bytree=0.8,
        min_child_samples=200,
        reg_lambda=1.0,
        random_state=random_state,
        n_jobs=-1,
        scale_pos_weight=scale_pos_weight,
    )

    model.fit(
        X_train, y_train,
        eval_set=[(X_valid, y_valid)],
        eval_metric="auc",
    )

    # 6) predict
    p_valid = model.predict_proba(X_valid)[:, 1]
    auc = roc_auc_score(y_valid, p_valid)

    # (A) threshold metrics (기본 threshold)
    y_pred = (p_valid >= threshold).astype(int)
    prec = precision_score(y_valid, y_pred, zero_division=0)
    rec  = recall_score(y_valid, y_pred, zero_division=0)
    f1   = f1_score(y_valid, y_pred, zero_division=0)

    # (B) Top-K (k=0 방지)
    n = len(p_valid)
    k = max(1, int(n * topk_ratio))
    idx = np.argsort(p_valid)[-k:]
    y_topk = np.zeros(n, dtype=int)
    y_topk[idx] = 1

    prec_k = precision_score(y_valid, y_topk, zero_division=0)
    rec_k  = recall_score(y_valid, y_topk, zero_division=0)
    f1_k   = f1_score(y_valid, y_topk, zero_division=0)

    # (C) perm AUC
    y_perm = np.random.permutation(y_valid)
    perm_auc = roc_auc_score(y_perm, p_valid)

    # ===== 추가: threshold sweep (F1 최대) =====
    best_t, best_f1, best_p, best_r = None, -1.0, None, None
    if sweep_thresholds:
        ts = np.linspace(t_min, t_max, t_steps)
        # 원래 너가 원한 형태(그대로): print(t, f1_score(...))
        print("\n=== threshold sweep (F1) ===")
        for t in ts:
            y_tmp = (p_valid >= t).astype(int)
            f1_tmp = f1_score(y_valid, y_tmp, zero_division=0)
            print(float(t), float(f1_tmp))

            if f1_tmp > best_f1:
                best_f1 = f1_tmp
                best_t = float(t)
                best_p = precision_score(y_valid, y_tmp, zero_division=0)
                best_r = recall_score(y_valid, y_tmp, zero_division=0)

    # 출력
    print(f"\nsplit_day={split_day} | train={len(train):,} valid={len(valid):,}")
    print(f"train pos_rate={y1/(y0+y1):.4%} | scale_pos_weight={scale_pos_weight:.2f}")
    print(f"ROC-AUC: {auc:.5f} | perm AUC: {perm_auc:.5f}")
    print(f"[Threshold={threshold:.2f}] Precision={prec:.4f} Recall={rec:.4f} F1={f1:.4f}")
    print(f"[Top {topk_ratio:.1%}] Precision={prec_k:.4f} Recall={rec_k:.4f} F1={f1_k:.4f} (k={k})")

    if sweep_thresholds and best_t is not None:
        print(f"\n[Best-F1 Threshold={best_t:.4f}] Precision={best_p:.4f} Recall={best_r:.4f} F1={best_f1:.4f}")

    return model, {
        "auc": auc,
        "perm_auc": perm_auc,
        "precision": prec,
        "recall": rec,
        "f1": f1,
        "precision_topk": prec_k,
        "recall_topk": rec_k,
        "f1_topk": f1_k,
        "k": k,
        # 추가 반환
        "best_threshold": best_t,
        "best_precision": best_p,
        "best_recall": best_r,
        "best_f1": best_f1,
    }, (X_valid, y_valid, p_valid)


# =========================
# 실행 코드
# =========================
bad_time = {"event_day_in_series", "event_week_in_series", "event_weekday_num"}
features_no_time = [c for c in features if c not in bad_time]

model2, m2, _ = train_eval_auc(
    panel_df=panel,
    features=features_no_time,
    valid_days=60,
    topk_ratio=0.01,
    threshold=0.5,
    sweep_thresholds=True,   # 여기 True면 스윕 찍힘
    t_min=0.01,
    t_max=0.30,
    t_steps=30
)

m2
```
